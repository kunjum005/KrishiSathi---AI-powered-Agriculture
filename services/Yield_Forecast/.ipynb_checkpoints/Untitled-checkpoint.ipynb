```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from tensorflow.keras.models import Sequential, load_model
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
import warnings
import joblib
import pickle
import sys
warnings.filterwarnings('ignore')

print("✅ All libraries imported successfully!")
print(f"TensorFlow version: {tf.__version__}")
print(f"Python version: {sys.version.split()[0]}")
```

```python
# Create realistic synthetic crop data for 40 Indian crops
np.random.seed(42)

def generate_crop_data():
    years = list(range(2010, 2024))
    
    # 40 Major Indian Crops
    crops = [
        # Cereals
        'Wheat', 'Rice', 'Maize', 'Barley', 'Sorghum', 'Pearl Millet', 'Finger Millet',
        # Pulses
        'Chickpea', 'Pigeon Pea', 'Green Gram', 'Black Gram', 'Lentil', 'Kidney Bean',
        # Oilseeds
        'Groundnut', 'Mustard', 'Soybean', 'Sunflower', 'Sesame', 'Castor', 'Linseed',
        # Cash Crops
        'Sugarcane', 'Cotton', 'Jute', 'Tobacco', 'Tea', 'Coffee', 'Rubber',
        # Spices
        'Black Pepper', 'Cardamom', 'Turmeric', 'Ginger', 'Coriander', 'Cumin', 'Fennel',
        # Fruits
        'Mango', 'Banana', 'Apple', 'Orange', 'Grapes', 'Pomegranate',
        # Vegetables
        'Potato', 'Tomato', 'Onion', 'Brinjal', 'Cabbage', 'Cauliflower'
    ]
    
    regions = ['North', 'South', 'East', 'West']
    
    data = []
    for year in years:
        for crop in crops:
            for region in regions:
                # Base yield by crop type (tons/hectare) - realistic for Indian conditions
                base_yield = {
                    # Cereals
                    'Wheat': 2.8, 'Rice': 3.5, 'Maize': 2.5, 'Barley': 2.2, 'Sorghum': 1.8, 
                    'Pearl Millet': 1.5, 'Finger Millet': 1.2,
                    # Pulses
                    'Chickpea': 1.1, 'Pigeon Pea': 0.9, 'Green Gram': 0.8, 'Black Gram': 0.7,
                    'Lentil': 0.9, 'Kidney Bean': 0.8,
                    # Oilseeds
                    'Groundnut': 1.5, 'Mustard': 1.4, 'Soybean': 1.8, 'Sunflower': 1.2,
                    'Sesame': 0.6, 'Castor': 1.3, 'Linseed': 0.8,
                    # Cash Crops
                    'Sugarcane': 65.0, 'Cotton': 1.2, 'Jute': 2.2, 'Tobacco': 1.8,
                    'Tea': 2.0, 'Coffee': 0.8, 'Rubber': 1.5,
                    # Spices
                    'Black Pepper': 0.5, 'Cardamom': 0.3, 'Turmeric': 0.7, 'Ginger': 0.9,
                    'Coriander': 0.8, 'Cumin': 0.6, 'Fennel': 0.7,
                    # Fruits
                    'Mango': 8.0, 'Banana': 25.0, 'Apple': 10.0, 'Orange': 12.0,
                    'Grapes': 15.0, 'Pomegranate': 9.0,
                    # Vegetables
                    'Potato': 20.0, 'Tomato': 25.0, 'Onion': 16.0, 'Brinjal': 18.0,
                    'Cabbage': 22.0, 'Cauliflower': 15.0
                }[crop]
                
                # Regional modifiers for India
                region_modifier = {
                    'North': 1.0, 'South': 1.1, 'East': 0.9, 'West': 0.95
                }[region]
                
                # Generate realistic features
                temperature = np.random.normal(25, 3) + np.random.choice([-2, 0, 2])
                rainfall = np.random.normal(800, 150) * (1 + 0.1 * np.sin(year - 2010))
                humidity = np.random.normal(65, 8)
                soil_nitrogen = np.random.normal(150, 25)
                soil_ph = np.random.uniform(5.5, 7.5)
                
                # Calculate yield with trend, seasonality, and noise
                trend = (year - 2010) * 0.02
                seasonality = 0.1 * np.sin(2 * np.pi * (year - 2010) / 5)
                weather_effect = np.random.normal(0, 0.1)
                
                yield_value = (base_yield + trend + seasonality + weather_effect) * region_modifier
                yield_value = max(yield_value, 0.1)  # Ensure positive yield
                
                data.append({
                    'year': year,
                    'crop': crop,
                    'region': region,
                    'temperature': temperature,
                    'rainfall': rainfall,
                    'humidity': humidity,
                    'soil_nitrogen': soil_nitrogen,
                    'soil_ph': soil_ph,
                    'yield': yield_value
                })
    
    return pd.DataFrame(data)

# Generate and save data
crop_data = generate_crop_data()
crop_data.to_csv('models/data/synthetic_crop_data.csv', index=False)

print("✅ Dataset created and saved in models/data/")
print(f"Dataset shape: {crop_data.shape}")
print(f"Total crops: {len(crop_data['crop'].unique())}")
print(f"Total records: {len(crop_data):,}")

print(f"\n📊 Crop Categories:")
categories = {
    'Cereals': ['Wheat', 'Rice', 'Maize', 'Barley', 'Sorghum', 'Pearl Millet', 'Finger Millet'],
    'Pulses': ['Chickpea', 'Pigeon Pea', 'Green Gram', 'Black Gram', 'Lentil', 'Kidney Bean'],
    'Oilseeds': ['Groundnut', 'Mustard', 'Soybean', 'Sunflower', 'Sesame', 'Castor', 'Linseed'],
    'Cash Crops': ['Sugarcane', 'Cotton', 'Jute', 'Tobacco', 'Tea', 'Coffee', 'Rubber'],
    'Spices': ['Black Pepper', 'Cardamom', 'Turmeric', 'Ginger', 'Coriander', 'Cumin', 'Fennel'],
    'Fruits': ['Mango', 'Banana', 'Apple', 'Orange', 'Grapes', 'Pomegranate'],
    'Vegetables': ['Potato', 'Tomato', 'Onion', 'Brinjal', 'Cabbage', 'Cauliflower']
}

for category, crop_list in categories.items():
    print(f"   • {category}: {len(crop_list)} crops")

print("\nFirst 5 rows:")
display(crop_data.head())
```

```python
class YieldLSTMForecaster:
    def __init__(self, lookback=3):
        self.lookback = lookback
        self.model = None
        self.scaler = MinMaxScaler()
        self.model_dir = 'models/saved_models'
        
    def prepare_crop_data(self, data, crop_type, region=None):
        """Prepare data for specific crop and optionally region"""
        if region:
            crop_data = data[(data['crop'] == crop_type) & (data['region'] == region)]
        else:
            crop_data = data[data['crop'] == crop_type]
            
        crop_data = crop_data.sort_values('year')
        
        if len(crop_data) < self.lookback + 1:
            print(f"⚠️ Insufficient data for {crop_type} in {region if region else 'all regions'}")
            return None, None, None
            
        features = ['temperature', 'rainfall', 'humidity', 'soil_nitrogen', 'yield']
        sequence_data = crop_data[features].values
        
        scaled_data = self.scaler.fit_transform(sequence_data)
        
        X, y = [], []
        for i in range(self.lookback, len(scaled_data)):
            X.append(scaled_data[i-self.lookback:i, :])
            y.append(scaled_data[i, -1])
            
        return np.array(X), np.array(y), self.scaler
    
    def build_model(self, input_shape):
        """Build simplified LSTM model architecture"""
        model = Sequential([
            LSTM(50, return_sequences=True, input_shape=input_shape),
            Dropout(0.2),
            LSTM(25, return_sequences=False),
            Dropout(0.2),
            Dense(16, activation='relu'),
            Dense(1)
        ])
        
        # Use string identifiers for loss and metrics to avoid serialization issues
        model.compile(
            optimizer='adam',
            loss='mse',  # Use string instead of tf.keras.losses.MSE
            metrics=['mae']  # Use string instead of tf.keras.metrics.MAE
        )
        
        return model
    
    def train_and_save_model(self, data, crop_type, region=None, epochs=50):
        """Train LSTM model and save it with compatible format"""
        X, y, scaler = self.prepare_crop_data(data, crop_type, region)
        
        if X is None or len(X) == 0:
            print(f"❌ Cannot train model for {crop_type} - insufficient data")
            return None
            
        self.scaler = scaler
        
        # Build and train model
        self.model = self.build_model((X.shape[1], X.shape[2]))
        
        print(f"🚀 Training model for {crop_type}...")
        print(f"📊 Training data shape: {X.shape}")
        
        history = self.model.fit(
            X, y,
            epochs=epochs,
            batch_size=8,
            validation_split=0.2,
            verbose=1,
            shuffle=False
        )
        
        # Save model and scaler with safe format
        model_filename = f"{crop_type.lower()}_lstm_model.h5"
        scaler_filename = f"{crop_type.lower()}_scaler.pkl"
        weights_filename = f"{crop_type.lower()}.weights.h5"  # FIXED: Added .weights. prefix
        config_filename = f"{crop_type.lower()}_model_config.pkl"
        
        # ALWAYS CREATE BACKUP FILES regardless of whether main save works
        print("💾 Creating backup model files (weights + config)...")
        
        # Save weights with correct extension
        self.model.save_weights(f'{self.model_dir}/{weights_filename}')
        
        # Save model configuration
        model_config = self.model.get_config()
        with open(f'{self.model_dir}/{config_filename}', 'wb') as f:
            pickle.dump(model_config, f)
        
        # Try to save the main model file (may fail due to compatibility issues)
        try:
            self.model.save(f'{self.model_dir}/{model_filename}')
            print(f"💾 Main model saved: {self.model_dir}/{model_filename}")
        except Exception as e:
            print(f"⚠️ Main model save failed (but backups created): {e}")
        
        # Save scaler using pickle
        with open(f'{self.model_dir}/{scaler_filename}', 'wb') as f:
            pickle.dump(scaler, f)
        
        print(f"💾 Scaler saved: {self.model_dir}/{scaler_filename}")
        print(f"💾 Backup weights saved: {self.model_dir}/{weights_filename}")
        print(f"💾 Backup config saved: {self.model_dir}/{config_filename}")
        
        return history
    
    def load_model(self, crop_type):
        """Load pre-trained model and scaler with compatibility handling"""
        model_filename = f"{crop_type.lower()}_lstm_model.h5"
        scaler_filename = f"{crop_type.lower()}_scaler.pkl"
        weights_filename = f"{crop_type.lower()}.weights.h5"  # FIXED: Updated to match new extension
        config_filename = f"{crop_type.lower()}_model_config.pkl"
        
        # First try to load the main model file
        try:
            self.model = load_model(f'{self.model_dir}/{model_filename}')
            print(f"✅ Loaded pre-trained model for {crop_type}")
            
        except Exception as e:
            print(f"⚠️ Standard load failed: {e}")
            print("🔄 Loading from backup files...")
            
            # Load from backup files (weights + config)
            try:
                # Load model configuration
                with open(f'{self.model_dir}/{config_filename}', 'rb') as f:
                    model_config = pickle.load(f)
                
                # Rebuild model from config
                self.model = Sequential.from_config(model_config)
                
                # Recompile the model
                self.model.compile(
                    optimizer='adam',
                    loss='mse',
                    metrics=['mae']
                )
                
                # Load weights
                self.model.load_weights(f'{self.model_dir}/{weights_filename}')
                print(f"✅ Loaded model from backup for {crop_type}")
                
            except Exception as e2:
                print(f"❌ Backup loading also failed: {e2}")
                return False
        
        # Load scaler
        try:
            with open(f'{self.model_dir}/{scaler_filename}', 'rb') as f:
                self.scaler = pickle.load(f)
            return True
        except Exception as e:
            print(f"❌ Error loading scaler for {crop_type}: {e}")
            return False
    
    def predict_future(self, data, crop_type, region=None, future_years=1):
        """Predict yield for future years"""
        # Try to load pre-trained model first
        model_loaded = self.load_model(crop_type)
        
        if not model_loaded:
            # If no pre-trained model, train new one
            print(f"🔄 No pre-trained model found. Training new model for {crop_type}...")
            self.train_and_save_model(data, crop_type, region, epochs=30)
        else:
            print(f"✅ Using pre-trained model for {crop_type}")
        
        X, y, _ = self.prepare_crop_data(data, crop_type, region)
        
        if X is None or self.model is None:
            print(f"❌ Cannot make predictions for {crop_type} - no model or data")
            return None
            
        # Get the last sequence from historical data
        last_sequence = X[-1]
        predictions = []
        current_sequence = last_sequence.copy()
        
        for year in range(future_years):
            next_pred = self.model.predict(current_sequence.reshape(1, self.lookback, X.shape[2]), verbose=0)
            
            # Create new data point with slight variations
            new_row = current_sequence[-1].copy()
            new_row[:-1] += np.random.normal(0, 0.01, len(new_row)-1)  # Small random changes
            new_row[-1] = next_pred[0, 0]  # Update yield prediction
            
            # Update sequence (remove oldest, add new prediction)
            current_sequence = np.vstack([current_sequence[1:], new_row])
            
            # Convert back to original scale
            temp_array = np.zeros((1, X.shape[2]))
            temp_array[0, -1] = next_pred[0, 0]
            actual_yield = self.scaler.inverse_transform(temp_array)[0, -1]
            
            predictions.append(max(actual_yield, 0.1))  # Ensure positive yield
            
        return predictions

# Initialize the forecaster
forecaster = YieldLSTMForecaster(lookback=3)
print("✅ LSTM Forecaster initialized successfully!")
```

```python
# Demonstrate predictions using the trained models for key crops
key_crops_to_predict = [
    'Wheat', 'Rice', 'Maize', 'Sugarcane', 'Cotton',
    'Groundnut', 'Mustard', 'Potato', 'Tomato', 'Mango'
]

print("🔮 MAKING PREDICTIONS FOR KEY INDIAN CROPS")
print("="*60)

for crop in key_crops_to_predict:
    print(f"\n🌱 {crop.upper()} PREDICTIONS")
    print("-" * 40)
    
    # This will use the trained model to make predictions
    predictions = forecaster.predict_future(crop_data, crop, future_years=2)
    
    if predictions:
        current_year = crop_data['year'].max()
        
        # Get historical data for comparison
        crop_hist = crop_data[crop_data['crop'] == crop]
        last_actual_yield = crop_hist[crop_hist['year'] == current_year]['yield'].mean()
        hist_avg = crop_hist['yield'].mean()
        
        print(f"📊 Historical Performance:")
        print(f"   • Last year ({current_year}): {last_actual_yield:.2f} tons/hectare")
        print(f"   • Historical average: {hist_avg:.2f} tons/hectare")
        
        print(f"\n🎯 Future Predictions:")
        for i, pred in enumerate(predictions, 1):
            year = current_year + i
            change_from_last = ((pred - last_actual_yield) / last_actual_yield) * 100
            print(f"   {year}: {pred:.2f} tons/hectare ({change_from_last:+.1f}%)")
        
        # Generate farmer recommendations
        avg_prediction = np.mean(predictions)
        print(f"\n💡 FARMER RECOMMENDATIONS:")
        
        if avg_prediction > hist_avg * 1.1:
            print("   🟢 EXCELLENT OUTLOOK")
            print("   • Increase storage capacity")
            print("   • Plan early market sales")
            print("   • Maintain optimal fertilizer levels")
        elif avg_prediction > hist_avg * 0.9:
            print("   🟡 NORMAL OUTLOOK") 
            print("   • Standard fertilizer application")
            print("   • Regular soil moisture monitoring")
        else:
            print("   🔴 CAUTION ADVISED")
            print("   • Increase soil nutrients")
            print("   • Implement water conservation")
            
    else:
        print(f"❌ Could not generate predictions for {crop}")

print(f"\n✅ All predictions completed for {len(key_crops_to_predict)} key crops!")
```

```python
# User-friendly interface for making predictions
print("🎯 CROP YIELD PREDICTION INTERFACE")
print("="*50)

def predict_yield_for_crop():
    """Simple interface to get predictions for any crop"""
    
    print("\nAvailable crops (40 Indian crops):")
    categories = {
        'Cereals': ['Wheat', 'Rice', 'Maize', 'Barley', 'Sorghum', 'Pearl Millet', 'Finger Millet'],
        'Pulses': ['Chickpea', 'Pigeon Pea', 'Green Gram', 'Black Gram', 'Lentil', 'Kidney Bean'],
        'Oilseeds': ['Groundnut', 'Mustard', 'Soybean', 'Sunflower', 'Sesame', 'Castor', 'Linseed'],
        'Cash Crops': ['Sugarcane', 'Cotton', 'Jute', 'Tobacco', 'Tea', 'Coffee', 'Rubber'],
        'Spices': ['Black Pepper', 'Cardamom', 'Turmeric', 'Ginger', 'Coriander', 'Cumin', 'Fennel'],
        'Fruits': ['Mango', 'Banana', 'Apple', 'Orange', 'Grapes', 'Pomegranate'],
        'Vegetables': ['Potato', 'Tomato', 'Onion', 'Brinjal', 'Cabbage', 'Cauliflower']
    }
    
    # Create case-insensitive mapping
    crop_mapping = {}
    all_crops = []
    for category, crops in categories.items():
        print(f"   {category}: {', '.join(crops)}")
        all_crops.extend(crops)
        for crop in crops:
            crop_mapping[crop.lower()] = crop  # Map lowercase to correct case
    
    while True:
        print(f"\n{'─'*50}")
        print("📝 PREDICTION INPUT FORM")
        print(f"{'─'*50}")
        
        # Get crop input (case-insensitive)
        crop_input = input("\nEnter crop name (or 'quit' to exit): ").strip()
        
        if crop_input.lower() == 'quit':
            print("👋 Thank you for using the Yield Prediction System!")
            break
            
        # Convert to correct case
        crop_name = crop_mapping.get(crop_input.lower())
        if not crop_name:
            print(f"❌ Crop '{crop_input}' not found in database.")
            print("💡 Please check the spelling and try again.")
            continue
        
        print(f"✅ Selected crop: {crop_name}")
        
        # Get region input (case-insensitive)
        print("\nAvailable regions: North, South, East, West")
        region_input = input("Enter region: ").strip().title()
        region_mapping = {
            'north': 'North', 'south': 'South', 'east': 'East', 'west': 'West'
        }
        region = region_mapping.get(region_input.lower(), 'North')  # Default to North
        print(f"✅ Selected region: {region}")
        
        # Get prediction years
        try:
            years_input = input("Enter number of years to predict (1-3): ").strip()
            years = int(years_input) if years_input else 1
            years = max(1, min(3, years))  # Limit between 1-3
            print(f"✅ Predicting for {years} year{'s' if years > 1 else ''}")
        except:
            years = 1
            print(f"⚠️  Using default: {years} year")
        
        print(f"\n⏳ Making predictions for {crop_name} in {region} region...")
        
        # Make prediction
        try:
            predictions = forecaster.predict_future(crop_data, crop_name, region=region, future_years=years)
            
            if predictions:
                current_year = crop_data['year'].max()
                
                print(f"\n✅ PREDICTION RESULTS for {crop_name.upper()}")
                print(f"{'─'*40}")
                
                # Get historical data for comparison
                crop_hist = crop_data[(crop_data['crop'] == crop_name) & (crop_data['region'] == region)]
                if not crop_hist.empty:
                    last_actual = crop_hist[crop_hist['year'] == current_year]['yield'].mean()
                    hist_avg = crop_hist['yield'].mean()
                    
                    print(f"📊 Historical Data:")
                    print(f"   • Last year ({current_year}): {last_actual:.2f} tons/hectare")
                    print(f"   • Historical average: {hist_avg:.2f} tons/hectare")
                
                print(f"\n🎯 Future Predictions:")
                for i, pred in enumerate(predictions, 1):
                    year = current_year + i
                    if 'last_actual' in locals():
                        change = ((pred - last_actual) / last_actual) * 100
                        print(f"   {year}: {pred:.2f} tons/hectare ({change:+.1f}%)")
                    else:
                        print(f"   {year}: {pred:.2f} tons/hectare")
                
                # Generate recommendations
                avg_prediction = np.mean(predictions)
                print(f"\n💡 FARMER RECOMMENDATIONS:")
                
                if 'hist_avg' in locals():
                    if avg_prediction > hist_avg * 1.1:
                        print("   🟢 EXCELLENT OUTLOOK - Above average yield expected")
                        print("   • Increase storage capacity by 15-20%")
                        print("   • Plan early market sales for better prices")
                        print("   • Maintain current fertilizer levels")
                    elif avg_prediction > hist_avg * 0.9:
                        print("   🟡 NORMAL OUTLOOK - Average yield expected")
                        print("   • Follow standard fertilizer schedule")
                        print("   • Monitor soil moisture regularly")
                        print("   • Implement integrated pest management")
                    else:
                        print("   🔴 CAUTION - Below average yield expected")
                        print("   • Increase soil nutrient application")
                        print("   • Implement water conservation measures")
                        print("   • Consult agricultural expert")
                else:
                    print("   • Follow standard agricultural practices")
                    print("   • Monitor weather forecasts regularly")
                    print("   • Maintain soil health with organic matter")
                
                # Show confidence level
                if 'hist_avg' in locals():
                                    # Show confidence level - FIXED VERSION
                if 'hist_avg' in locals():
                    # More realistic confidence calculation
                    base_confidence = 75  # Start with medium confidence
                    
                    # Adjust based on crop type (you can expand this)
                    high_confidence_crops = ['Wheat', 'Rice', 'Maize', 'Sugarcane', 'Cotton']
                    medium_confidence_crops = ['Groundnut', 'Mustard', 'Potato', 'Tomato', 'Mango', 'Banana']
                    
                    if crop_name in high_confidence_crops:
                        base_confidence += 15
                    elif crop_name in medium_confidence_crops:
                        base_confidence += 10
                    
                    # Adjust based on prediction stability
                    if len(predictions) > 1:
                        variation = np.std(predictions) / np.mean(predictions)
                        if variation < 0.05:  # Low variation
                            base_confidence += 5
                        elif variation > 0.15:  # High variation
                            base_confidence -= 10
                    
                    # Final confidence (capped between 60-95%)
                    confidence = max(60, min(95, base_confidence))
                    print(f"\n📈 Prediction Confidence: {confidence:.0f}%")
                else:
                    print(f"\n📈 Prediction Confidence: 70% (estimated)")
                    print(f"\n📈 Prediction Confidence: {confidence:.0f}%")
                else:
                    print(f"\n📈 Prediction Confidence: 75% (estimated)")
                
            else:
                print(f"❌ Could not generate predictions for {crop_name}")
                
        except Exception as e:
            print(f"❌ Error making prediction: {e}")
            print("💡 The system will train a new model for this crop...")
            
            # Try to train and predict again
            try:
                print("🔄 Training model for this crop...")
                predictions = forecaster.predict_future(crop_data, crop_name, region=region, future_years=years)
                if predictions:
                    print(f"✅ Success! Prediction: {predictions[0]:.2f} tons/hectare")
                else:
                    print("❌ Still unable to generate prediction")
            except:
                print("❌ Failed to generate prediction for this crop")
        
        print(f"\n{'─'*50}")
        print("🔄 Ready for next prediction...")

# Run the prediction interface
predict_yield_for_crop()
```
